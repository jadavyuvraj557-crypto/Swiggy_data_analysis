```python
rating_by_cuisine = (
    df.groupby("Cuisine")
      .agg(
          Average_Rating=("Rating", "mean"),
          Total_Restaurants=("Restaurant Name", "count")
      )
)

rating_by_cuisine = rating_by_cuisine[
    rating_by_cuisine["Total_Restaurants"] >= 50
]

rating_by_cuisine = rating_by_cuisine.sort_values(
    by="Average_Rating",
    ascending=False
)

rating_by_cuisine.head(10)
```
